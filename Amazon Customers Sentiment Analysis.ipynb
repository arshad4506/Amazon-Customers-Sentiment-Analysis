```python
import pandas as pd
import numpy as np

data = {
    "Review_ID": range(1, 31),

    "Product": [
        "Wireless Earbuds", "Wireless Earbuds", "Smart Watch",
        "Smart Watch", "Laptop Stand", "Laptop Stand",
        "Bluetooth Speaker", "Bluetooth Speaker", "Mobile Charger",
        "Mobile Charger", "Coffee Maker", "Coffee Maker",
        "Backpack", "Backpack", "Keyboard",
        "Keyboard", "Mouse", "Mouse", "Power Bank", "Power Bank",
        "Headphones", "Headphones", "Water Bottle", "Water Bottle",
        "Phone Case", "Phone Case", "USB Cable", "USB Cable",
        "Desk Lamp", "Desk Lamp"
    ],

    "Rating": [
        5, 4, 5, 2, 4, 3, 5, 1, 5, 2,
        4, 3, 5, 2, 4, 5, 4, 1, 5, 2,
        5, 2, 4, 3, 5, 1, 4, 2, 5, 3
    ],

    "Review": [
        "Excellent product and very good sound quality",
        "Good earbuds and battery life is impressive",
        "Amazing smartwatch with useful features",
        "The watch stopped working after a few days",
        "Very useful and strong laptop stand",
        "The stand is okay but could be stronger",
        "Excellent speaker with clear sound",
        "Very poor sound quality and stopped working",
        "Fast charging and excellent performance",
        "Charging is very slow and disappointing",
        "Makes great coffee and easy to use",
        "The coffee maker is average",
        "Very comfortable backpack with enough space",
        "The bag quality is poor and the zip broke",
        "Keyboard feels smooth and works perfectly",
        "I really love this keyboard",
        "Good mouse with accurate movement",
        "Mouse stopped working suddenly",
        "Excellent power bank with long battery life",
        "Battery drains very quickly",
        "Amazing headphones with excellent sound",
        "Sound quality is terrible and uncomfortable",
        "Good quality water bottle and easy to carry",
        "The bottle is okay but not very impressive",
        "Excellent phone case and perfect fit",
        "The case broke after one week",
        "Good USB cable and fast charging",
        "Cable stopped working after a few days",
        "Excellent desk lamp with bright light",
        "The lamp is okay but brightness could be better"
    ]
}

df = pd.DataFrame(data)

print("Dataset created successfully")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
```

```python
def get_sentiment(rating):
    if rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

df["Sentiment"] = df["Rating"].apply(get_sentiment)

print(df.head(10))
```

```python
print("Customer Analysis")
print("==================")

print("\n1. Total Customers:")
print(df["Customer_ID"].nunique())

print("\n2. Average Customer Age:")
print(df["Age"].mean())

print("\n3. Average Purchase Amount:")
print(df["Purchase_Amount"].mean())

print("\n4. Highest Purchase Amount:")
print(df["Purchase_Amount"].max())

print("\n5. Lowest Purchase Amount:")
print(df["Purchase_Amount"].min())

print("\n6. Average Rating:")
print(df["Rating"].mean())
```
